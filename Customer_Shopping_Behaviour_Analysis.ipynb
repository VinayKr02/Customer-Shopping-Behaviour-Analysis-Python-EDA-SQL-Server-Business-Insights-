# Feature Engineering

```python
# add columns in the data set
labels=['Young','Adult','Midde_Aged','Senior']
cs['age_group']=pd.qcut(cs['age'],q=4,labels=labels)
```

```python
cs[['age','age_group']].head()
```

```python
#add column frequency purchase days
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,'Monthly':30,'Quarterly':90,'Bi-Weekly':14,'Annually':365,'Every 3 Months':90}

cs['frequency_of_purchases_days']=cs['frequency_of_purchases'].map(frequency_mapping)
```

# Graphical Representation of data set

```python
#How many customers belong to each gender?
import seaborn as sns

plt.figure(figsize=(6,4))

ax = sns.countplot(data=cs, x='gender',color='blue')

for container in ax.containers:
    ax.bar_label(container)

plt.title('Customer Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Customers')
plt.show()
```

```python
#Which individual items are most popular?
top_items = cs['item_purchased'].value_counts().head(10)

plt.figure(figsize=(8,5))

ax = sns.barplot(
    x=top_items.values,
    y=top_items.index
)

for container in ax.containers:
    ax.bar_label(container)

plt.title('Top 10 Most Purchased Items')
plt.xlabel('Number of Purchases')
plt.ylabel('Item')
plt.show()
```

```python
plt.figure(figsize=(8,4))

ax=sns.histplot(
    data=cs,
    x='purchase_amount',
    bins=20,
    kde=True,
     color='#55A868'
)

for container in ax.containers:
    ax.bar_label(container)

plt.title('Distribution of Purchase Amount')
plt.xlabel('Purchase Amount')
plt.ylabel('Number of Customers')
plt.show()
```

```python
"""Purchase amount by category"""

plt.figure(figsize=(8,5))

sns.boxplot(
    data=cs,
    x='category',
    y='purchase_amount'
)

plt.title('Purchase Amount by Category')
plt.xlabel('Category')
plt.ylabel('Purchase Amount')
plt.xticks(rotation=30)
plt.show()
```

```python
avg_purchase = (
    cs.groupby('category')['purchase_amount']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))

ax = sns.barplot(
    x=avg_purchase.values,
    y=avg_purchase.index
)

for container in ax.containers:
    ax.bar_label(container, fmt='%.1f')

plt.title('Average Purchase Amount by Category')
plt.xlabel('Average Purchase Amount')
plt.ylabel('Category')
plt.show()
```

```python
plt.figure(figsize=(10,5))

ax = sns.countplot(
    data=cs,
    x='age_group',
    hue='category'
)

for container in ax.containers:
    ax.bar_label(container)

plt.title('Product Category Distribution by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Purchases')
plt.xticks(rotation=30)

plt.legend(
    title='Category',
    fontsize=8,
    title_fontsize=9
)

plt.show()
```

```python
order = ['Weekly','Bi-Weekly','Fortnightly','Monthly','Quarterly','Annually']
order = [o for o in order if o in cs['frequency_of_purchases'].unique()]

fig, ax = plt.subplots(figsize=(8,5))
sns.countplot(data=cs, y='frequency_of_purchases', order=order, palette='viridis', ax=ax)
ax.set_title('How Often Customers Purchase')
ax.set_xlabel('Number of Customers')
ax.set_ylabel('Purchase Frequency')
plt.tight_layout()
plt.show()
```
